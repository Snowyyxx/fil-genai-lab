# Hybrid Search + RRF + Cross-Encoder Rerank

**The two biggest retrieval wins, on top of the Day-1 bench.**

Dense (vector) retrieval is great at *meaning* but blind to *exact tokens* — error
codes, IDs, SKUs, function names. And because everything near the query in vector
space looks similar, it happily returns **near-duplicates** instead of the one doc
you needed. Two techniques fix most of this:

1. **Hybrid search + Reciprocal Rank Fusion (RRF)** — run **BM25** (lexical, nails
   exact terms) *and* **dense** (semantic) retrieval, then fuse their two ranked
   lists by rank position. No score normalization headaches.
2. **Retrieve-wide-then-rerank** — pull a wide candidate set (top-N), then score
   each *(query, doc)* pair with a **cross-encoder** (`bge-reranker`). Slow but
   accurate; only runs on N candidates, not the whole corpus.

We reuse the **exact retrieval metrics from Day 1** (`rag_eval/`) — hit@k, MRR,
precision@k, recall@k, ndcg@k — and record the **recall + precision@5 lift** of
`dense-only → hybrid+RRF → hybrid+rerank`.

Everything runs on CPU. First run downloads ~200 MB of model weights
(MiniLM embedder + bge-reranker-base).

> Resources: [Weaviate — hybrid search & fusion](https://weaviate.io/blog/hybrid-search-explained) ·
> [Cohere — Rerank / cross-encoders](https://docs.cohere.com/docs/reranking)

## 1. A corpus built to break dense-only retrieval

The Day-1 corpus was trivia — clean topical separation, so dense retrieval aced
it and there was nothing for hybrid/rerank to fix. To *see* the lift, we need a
corpus with the two failure modes from real support/knowledge bases:

- **Exact-term traps** — error codes (`E-4021`), ticket IDs (`TICKET-8842`),
  config keys (`max_retries`), version strings (`v2.3.1`). Embeddings smear these
  into "generic technical text"; BM25 matches them character-for-character.
- **Near-duplicate clusters** — several docs that say almost the same thing.
  Dense retrieval fills its top-k with these look-alikes and starves the *other*
  relevant docs, tanking recall.

```python
DOCUMENTS = {
    # --- Auth cluster (semantically similar, but the exact error code matters) ---
    "d_e4021":   "Error E-4021 means the API token has expired. Generate a new token in Settings > API Keys and update your client.",
    "d_auth1":   "Authentication can fail when your credentials are no longer valid. Check that your login details are current.",
    "d_auth2":   "If sign-in is rejected, your session may have timed out. Re-authenticate to obtain a fresh session.",
    "d_auth3":   "Access denied errors usually indicate that the account lacks permission for the requested resource.",

    # --- Rate-limit cluster ---
    "d_e4029":   "Error E-4029 is returned when you exceed the rate limit of 100 requests per minute. Back off and retry with exponential delay.",
    "d_rate1":   "Sending too many requests in a short window will get your calls throttled by the server.",
    "d_rate2":   "The service caps how frequently a client may call the API to protect shared infrastructure.",

    # --- Config / retries ---
    "d_retries": "The max_retries setting controls how many times the SDK re-attempts a failed request. The default is 3; set it to 0 to disable retries.",
    "d_timeout": "The request_timeout option sets how long, in seconds, the client waits for a response before giving up. Default is 30.",

    # --- Billing ticket ---
    "d_ticket":  "TICKET-8842: customer reports duplicate charges on invoice INV-2024-0917. Root cause was a retried payment webhook. Resolved by idempotency key.",
    "d_billing": "Invoices are generated on the first of each month and charged to the payment method on file.",

    # --- Versioning ---
    "d_v231":    "Release v2.3.1 patches a memory leak in the connection pool and requires no migration. Upgrade with pip install --upgrade ourpkg.",
    "d_v230":    "Release v2.3.0 introduced the async client and changed the default request_timeout from 60 to 30 seconds.",

    # --- Distractors (unrelated topics, to make retrieval non-trivial) ---
    "d_python":  "Python is a high-level programming language created by Guido van Rossum and first released in 1991.",
    "d_everest": "Mount Everest is Earth's highest mountain above sea level, with a peak at 8,849 metres in the Himalayas.",
    "d_photo":   "Photosynthesis is the process by which green plants convert sunlight, water, and carbon dioxide into glucose and oxygen.",
    "d_water":   "Water is a chemical compound with the formula H2O: two hydrogen atoms bonded to one oxygen atom.",
    "d_paris":   "Paris is the capital and most populous city of France, with about 2.1 million people within the city limits.",
}

DOC_IDS  = list(DOCUMENTS.keys())
DOC_TEXT = list(DOCUMENTS.values())
print(f"{len(DOCUMENTS)} documents loaded")
```

## 2. Dense retriever — the Day-1 baseline

Identical to Day 1: MiniLM embeddings, cosine similarity, top-k. This is the bar
hybrid + rerank has to beat.

```python
import numpy as np
from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")            # ~90 MB
doc_embeddings = embedder.encode(DOC_TEXT, normalize_embeddings=True)
print("doc embedding matrix:", doc_embeddings.shape)

def dense_search(query, k=5):
    """Return top-k (doc_id, score) by cosine similarity, highest first."""
    q = embedder.encode([query], normalize_embeddings=True)[0]
    scores = doc_embeddings @ q
    order = np.argsort(-scores)[:k]
    return [(DOC_IDS[i], float(scores[i])) for i in order]

for did, s in dense_search("What does error E-4021 mean?", k=5):
    print(f"{s:.3f}  {did}")
```

Notice the dense results for `E-4021`: the exact-code doc often gets buried under
the *semantically* similar auth docs (`d_auth1`, `d_auth2`…). The embedder sees
"authentication problem" and can't tell `E-4021` from `E-4029`. That's the gap
BM25 closes.

## 3. BM25 — the lexical retriever

BM25 is the classic bag-of-words ranking function: it scores a doc by how many
**query terms it contains**, weighted by term rarity (IDF) and dampened for length.
It has zero notion of meaning — but that's exactly why `E-4021` matches `E-4021`
and nothing else.

```python
import re
from rank_bm25 import BM25Okapi

def tokenize(text):
    # Lowercase, but keep alphanumerics + hyphens together so codes like
    # 'e-4021', 'inv-2024-0917', 'v2.3.1', 'max_retries' survive as single tokens.
    return re.findall(r"[a-z0-9][a-z0-9._-]*", text.lower())

bm25 = BM25Okapi([tokenize(t) for t in DOC_TEXT])

def bm25_search(query, k=5):
    """Return top-k (doc_id, score) by BM25, highest first."""
    scores = bm25.get_scores(tokenize(query))
    order = np.argsort(-scores)[:k]
    return [(DOC_IDS[i], float(scores[i])) for i in order]

for did, s in bm25_search("What does error E-4021 mean?", k=5):
    print(f"{s:.3f}  {did}")
```

BM25 puts `d_e4021` at the top — the code is a rare token, so IDF rewards it
heavily. But BM25 is brittle the other way: a query phrased *without* the exact
words ("my login keeps getting rejected") scores near zero against docs that mean
the same thing with different vocabulary. **Dense covers BM25's weakness and vice
versa** — so we fuse them.

## 4. Reciprocal Rank Fusion (RRF)

How do you merge two ranked lists whose scores aren't comparable (cosine ∈ [-1,1]
vs unbounded BM25)? **Don't use the scores at all — use the ranks.** RRF gives each
doc a score of

$$\text{RRF}(d) = \sum_{\text{lists}} \frac{1}{k_{\text{rrf}} + \text{rank}_{\text{list}}(d)}$$

where `rank` is 1-indexed and `k_rrf` (conventionally 60) damps the contribution
of low ranks. A doc that ranks high in **either** retriever floats up; a doc that
ranks high in **both** wins outright. Simple, robust, no tuning.

```python
from collections import defaultdict

def rrf_fuse(ranked_lists, k_rrf=60):
    """ranked_lists: list of [(doc_id, score), ...] each already sorted best-first.
    Returns a single [(doc_id, rrf_score), ...] sorted best-first."""
    scores = defaultdict(float)
    for ranked in ranked_lists:
        for rank, (doc_id, _score) in enumerate(ranked, start=1):
            scores[doc_id] += 1.0 / (k_rrf + rank)
    return sorted(scores.items(), key=lambda x: -x[1])

def hybrid_search(query, k=5, pool=10, k_rrf=60):
    """Fuse dense + BM25 top-`pool` lists via RRF, return top-k (doc_id, rrf_score)."""
    dense = dense_search(query, k=pool)
    lexical = bm25_search(query, k=pool)
    return rrf_fuse([dense, lexical], k_rrf=k_rrf)[:k]

print("Query: 'my login keeps getting rejected'  (paraphrase — no exact terms)\n")
for did, s in hybrid_search("my login keeps getting rejected", k=5):
    print(f"{s:.4f}  {did}")
```

## 5. Cross-encoder reranker (retrieve-wide-then-rerank)

Both retrievers above score query and doc **independently** (a query vector, a doc
vector, one dot product) — fast, but the two never actually "read" each other. A
**cross-encoder** feeds `[query] [SEP] [doc]` through a transformer *together* and
outputs a single relevance score. Far more accurate, far too slow to run over a
whole corpus — so the pattern is:

1. **Retrieve wide** — get a candidate pool (here: hybrid top-`pool`).
2. **Rerank** — cross-encode every `(query, candidate)` pair and re-sort.
3. Keep the top-k.

We use `BAAI/bge-reranker-base` (~280 MB). Cohere Rerank is the hosted equivalent
— same idea, different weights.

```python
from sentence_transformers import CrossEncoder

reranker = CrossEncoder("BAAI/bge-reranker-base")             # ~280 MB

def rerank_search(query, k=5, pool=10, base=hybrid_search):
    """Retrieve a wide pool with `base`, then re-order it with the cross-encoder."""
    candidates = [doc_id for doc_id, _ in base(query, k=pool)]
    pairs = [(query, DOCUMENTS[doc_id]) for doc_id in candidates]
    ce_scores = reranker.predict(pairs)
    ranked = sorted(zip(candidates, ce_scores), key=lambda x: -x[1])
    return [(doc_id, float(s)) for doc_id, s in ranked[:k]]

print("Query: 'How do I stop the SDK from retrying failed requests?'\n")
for did, s in rerank_search("How do I stop the SDK from retrying failed requests?", k=5):
    print(f"{s:.3f}  {did}")
```

## 6. Retrieval metrics — reused verbatim from Day 1

Same deterministic implementations as `rag_eval/rag_evaluation.ipynb`, so the
numbers are directly comparable to the Day-1 bench. They grade a ranked list of
`retrieved` ids against a set of `relevant` gold ids.

```python
import math

def hit_rate(retrieved, relevant, k):
    relevant = set(relevant)
    return 1.0 if any(d in relevant for d in retrieved[:k]) else 0.0

def reciprocal_rank(retrieved, relevant):
    relevant = set(relevant)
    for i, d in enumerate(retrieved, start=1):
        if d in relevant:
            return 1.0 / i
    return 0.0

def precision_at_k(retrieved, relevant, k):
    relevant, topk = set(relevant), retrieved[:k]
    if not topk:
        return 0.0
    return sum(d in relevant for d in topk) / len(topk)

def recall_at_k(retrieved, relevant, k):
    relevant = set(relevant)
    if not relevant:
        return 0.0
    return len(set(retrieved[:k]) & relevant) / len(relevant)

def ndcg_at_k(retrieved, relevant, k):
    relevant, topk = set(relevant), retrieved[:k]
    dcg = sum((1.0 if d in relevant else 0.0) / math.log2(i + 1)
              for i, d in enumerate(topk, start=1))
    ideal = min(len(relevant), k)
    idcg = sum(1.0 / math.log2(i + 1) for i in range(1, ideal + 1))
    return dcg / idcg if idcg > 0 else 0.0

print("metrics ready")
```

## 7. The evaluation set

Each query has **multiple gold docs** (so precision@5 and recall@5 are meaningful)
and is deliberately chosen to stress one failure mode:

- exact-code queries (BM25 should help),
- paraphrased queries (dense should help),
- ambiguous queries where two similar codes/docs compete (rerank should help).

```python
EVAL_SET = [
    {"question": "What does error E-4021 mean and how do I fix it?",
     "relevant_ids": ["d_e4021", "d_auth1", "d_auth2"]},
    {"question": "I keep getting E-4029, what is going on?",
     "relevant_ids": ["d_e4029", "d_rate1", "d_rate2"]},
    {"question": "my login keeps getting rejected",
     "relevant_ids": ["d_auth1", "d_auth2", "d_auth3", "d_e4021"]},
    {"question": "How do I stop the SDK from retrying failed requests?",
     "relevant_ids": ["d_retries"]},
    {"question": "What changed in release v2.3.1?",
     "relevant_ids": ["d_v231"]},
    {"question": "customer sees duplicate charges on their invoice",
     "relevant_ids": ["d_ticket", "d_billing"]},
    {"question": "how long does the client wait before timing out a request?",
     "relevant_ids": ["d_timeout", "d_v230"]},
    {"question": "I am being throttled for sending too many requests",
     "relevant_ids": ["d_rate1", "d_rate2", "d_e4029"]},
]
print(len(EVAL_SET), "eval questions")
```

## 8. The bench — dense → hybrid+RRF → hybrid+rerank

Run all three retrievers over the eval set at **k=5** and average every metric.
`recall@5` and `prec@5` are the headline numbers the project asks us to track.

```python
import pandas as pd

K = 5
POOL = 10   # candidate width for hybrid fusion and reranking

RETRIEVERS = {
    "dense":          lambda q: [d for d, _ in dense_search(q, k=K)],
    "hybrid_rrf":     lambda q: [d for d, _ in hybrid_search(q, k=K, pool=POOL)],
    "hybrid_rerank":  lambda q: [d for d, _ in rerank_search(q, k=K, pool=POOL)],
}

def evaluate(retriever_fn):
    agg = {"hit@5": [], "mrr": [], "prec@5": [], "recall@5": [], "ndcg@5": []}
    for ex in EVAL_SET:
        retr = retriever_fn(ex["question"])
        rel = ex["relevant_ids"]
        agg["hit@5"].append(hit_rate(retr, rel, K))
        agg["mrr"].append(reciprocal_rank(retr, rel))
        agg["prec@5"].append(precision_at_k(retr, rel, K))
        agg["recall@5"].append(recall_at_k(retr, rel, K))
        agg["ndcg@5"].append(ndcg_at_k(retr, rel, K))
    return {m: round(float(np.mean(v)), 3) for m, v in agg.items()}

bench = pd.DataFrame({name: evaluate(fn) for name, fn in RETRIEVERS.items()}).T
bench.index.name = "retriever"
bench
```

### The lift — vs the dense baseline

The single number the project asks us to record: how much did each metric improve
over dense-only?

```python
lift = (bench - bench.loc["dense"]).round(3)
lift.columns = [f"\u0394 {c}" for c in lift.columns]
print("Absolute lift over dense-only (higher = better):")
lift
```

## 9. Per-query diagnosis

Averages hide *where* the win comes from. This table shows recall@5 per query for
each retriever — you should see the biggest jumps on the exact-code and
ambiguous-competing-code queries, which is exactly what hybrid + rerank targets.

```python
rows = []
for ex in EVAL_SET:
    row = {"question": ex["question"][:42]}
    for name, fn in RETRIEVERS.items():
        retr = fn(ex["question"])
        row[name] = round(recall_at_k(retr, ex["relevant_ids"], K), 2)
    rows.append(row)

pd.DataFrame(rows).set_index("question")
```

## 10. What the numbers tell you

- **Hybrid+RRF > dense on exact-term queries.** BM25 rescues docs the embedder
  couldn't distinguish (`E-4021` vs `E-4029`, `v2.3.1` vs `v2.3.0`). RRF folds
  that lexical signal in without needing to normalize incomparable scores.
- **Rerank > hybrid on ordering (MRR / ndcg / precision).** The cross-encoder
  reads query and doc *together*, so it breaks ties the bag-of-vectors stage
  can't — pushing the single best doc to rank 1. It only rescores `POOL`
  candidates, so the cost is bounded.
- **Recall is capped by the retrieve-wide stage.** The reranker can only reorder
  what the pool contained — if a gold doc never made the top-`POOL`, no reranker
  can recover it. Widen `POOL` to trade latency for recall.

### Knobs to try
- `POOL` (candidate width): larger → higher recall ceiling, slower rerank.
- `k_rrf` in RRF: smaller → sharper preference for top ranks.
- Weighted RRF: give BM25 or dense a multiplier if one is consistently stronger
  on your traffic.
- Swap `bge-reranker-base` for `bge-reranker-large` (stronger) or a hosted
  **Cohere Rerank** call (no local weights).

### Where this plugs into Day 1
Drop `rerank_search` in as the retriever inside the Day-1 `rag()` pipeline: better
context in → better Faithfulness / Context-Relevance / Answer-Correctness out. The
retrieval lift above is the leading indicator of that downstream generation gain.
