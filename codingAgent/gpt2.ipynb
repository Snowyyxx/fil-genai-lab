```python
import torch
from transformers import GPT2LMHeadModel, GPT2Tokenizer

def setup_model(model_name="gpt2"):
    """Loads the tokenizer and the model."""
    print(f"Loading {model_name}...")
    # 'gpt2' is the 124M parameter base model. 
    # You can also use 'gpt2-medium' (355M), 'gpt2-large' (774M), or 'gpt2-xl' (1.5B)
    tokenizer = GPT2Tokenizer.from_pretrained(model_name)
    model = GPT2LMHeadModel.from_pretrained(model_name)
    
    # Set the pad token to avoid warning messages during generation
    tokenizer.pad_token_id = tokenizer.eos_token_id
    
    return tokenizer, model

def generate_text(prompt, tokenizer, model, max_length=100):
    """Generates text based on the provided prompt."""
    # Convert the text string into token IDs
    inputs = tokenizer.encode(prompt, return_tensors="pt")

    # Generate the output tokens
    with torch.no_grad(): # Disable gradient calculation for faster inference
        outputs = model.generate(
            inputs,
            max_length=max_length,
            num_return_sequences=1,
            no_repeat_ngram_size=2,    # Prevents looping phrases
            temperature=0.7,           # Controls randomness (higher = more creative)
            top_p=0.9,                 # Nucleus sampling for better quality
            do_sample=True,            # Required to use temperature and top_p
            pad_token_id=tokenizer.eos_token_id
        )

    # Decode the token IDs back into a human-readable string
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

if __name__ == "__main__":
    # Initialize
    tokenizer, model = setup_model("gpt2")
    
    # Run Inference
    prompt = "The architecture of modern data pipelines requires"
    print(f"\nPrompt: {prompt}")
    print("-" * 40)
    
    result = generate_text(prompt, tokenizer, model, max_length=75)
    print(result)
```
